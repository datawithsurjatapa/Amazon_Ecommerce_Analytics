```python
import pandas as pd
import numpy as np

df = pd.read_csv("Amazon_Ecommerce.csv")

print(df.shape)
df.head()
```

```python
#Data type checking
df.info()
```

```python
df['purchase_date'] = pd.to_datetime(df['purchase_date'])
```

```python
df.info()
```

```python
# Missing values checking
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
#Duplicate IDs Checking
print(df['user_id'].nunique())
print(df['product_id'].nunique())
print(df['seller_id'].nunique())
```

```python
#Validate Numerical Data

df[['price', 'final_price']].describe()
```

```python
(df['price'] <= 0).sum()
(df['final_price'] <= 0).sum()
```

```python
#Statistical summary
df.describe()
```

```python
#Shipping time
df['shipping_time_days'].value_counts().sort_index()
```

```python
#Categorical Consistency Checking
for col in ['category', 'subcategory', 'brand',
            'location', 'device', 'payment_method',
            'delivery_status']:
    print("\n", col)
    print(df[col].unique())
```

```python
#Delivery Status vs Return Status
pd.crosstab(
    df['is_returned'],
    df['delivery_status']
)
```

```python
#Category analysis
df.groupby('category')['final_price'].sum().sort_values(ascending=False)
```

```python
df['category'].value_counts()
```

```python
#Customer analysis
customer_sales = (
    df.groupby('user_id')['final_price']
      .sum()
      .sort_values(ascending=False)
)

customer_sales.head(10)
```

```python
#Seller analysis
seller_sales = (
    df.groupby('seller_id')['final_price']
      .sum()
      .sort_values(ascending=False)
)

seller_sales.head(10)
```

```python
#Time-based analysis
monthly_revenue = (
    df.groupby(df['purchase_date'].dt.to_period('M'))['final_price']
      .sum()
)

monthly_revenue
```
